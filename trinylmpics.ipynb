```python
import pandas as pd
import matplotlib.pyplot as plt 
import numpy as np 

df = pd.read_csv("trinity_xc_data.csv")
df.head()
```

```python
df.columns
```

```python
df = df.rename(columns={
    'Name':'name',
    'Were you born in Texas?':'texas_born',
    'Do you have any claim to being out of state?': 'oos_claim',
    'If so, how?': 'reasoning',
    'Which sports would you want to play? \r\n(Obviously not binding)': 'sports'
})
df.columns
```

```python
df = df.drop(columns='Timestamp')
```

```python
df['texas_born'] = df['texas_born'].str.lower().map({'yes': 1, 'no':0})
df['oos_claim'] = df['oos_claim'].str.lower().map({'yes':1, 'no':0})
```

```python
sport_dummies = df['sports'].str.replace(r',\s*', ',', regex=True).str.get_dummies(sep=',')
sport_dummies = sport_dummies.rename(columns={
    'Football': 'football',
    'Basketball': 'basketball',
    'Soccer': 'soccer',
    'Kickball':'kickball',
    'Ultimate Frisbee': 'frisbee',
    'Volleyball': 'volleyball'
})
sports_cols = ['football', 'basketball', 'soccer', 'kickball', 'frisbee', 'volleyball']
df[sports_cols] = sport_dummies[sports_cols]
```

```python
df = df.drop(columns='sports')
```

```python
df[['first_name', 'last_name']] = df['name'].str.split(' ', n=1, expand=True)
df=df.drop(columns=['name', 'last_name'])
cols = ['first_name'] + [c for c in df.columns if c != 'first_name']
df = df[cols]
```

```python
df['reasoning']=df['reasoning'].fillna(0)
df['reasoning'] = df['reasoning'].apply(lambda x: 1 if x!= 0 else 0)
```

```python
df['sports'] = df[sports_cols].sum(axis=1)
```

```python
df_football = df[df['football'] == 1]
df_football = df_football.drop(columns=['basketball', 'soccer', 'kickball', 'frisbee', 'volleyball','sports'])
df_football['either'] = df_football.sum(axis=1, numeric_only=True) - 3
df_football = df_football.sort_values('either', ascending=False)
df_football
```

```python
df_basketball = df[df['basketball'] == 1]
df_basketball.drop(columns=[sports_cols]) #this doesn't work for some reason, I can't figure it out rn
df_basketball['either'] = df_basketball.sum(axis=1, numeric_only=True) - 2
df_basketball = df_basketball.sort_values('either', ascending=False)
df_basketball
```
