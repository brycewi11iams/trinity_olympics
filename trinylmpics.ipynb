```python
import pandas as pd
import matplotlib.pyplot as plt 
import numpy as np 

df = pd.read_csv("trinity_xc_data.csv")
df
```

```python
df.columns
```

```python
df = df.rename(columns={
    'Name':'name',
    'Were you born in Texas?':'texas_born',
    'Do you have any claim to being out of state?': 'oos_claim',
    'If so, how?': 'reasoning',
    'Which sports would you want to play? \n(Obviously not binding)': 'sports'
})
df.columns
```

```python
df = df.drop(columns='Timestamp')
```

```python
df['texas_born'] = df['texas_born'].str.lower().map({'yes': 1, 'no':0})
df['oos_claim'] = df['oos_claim'].str.lower().map({'yes':1, 'no':0})
```

```python
sport_dummies = df['sports'].str.replace(r',\s*', ',', regex=True).str.get_dummies(sep=',')
sport_dummies = sport_dummies.rename(columns={
    'Football': 'football',
    'Basketball': 'basketball',
    'Soccer': 'soccer',
    'Kickball':'kickball',
    'Ultimate Frisbee': 'frisbee',
    'Volleyball': 'volleyball'
})
sports_cols = ['football', 'basketball', 'soccer', 'kickball', 'frisbee', 'volleyball']
df[sports_cols] = sport_dummies[sports_cols]

df
```

```python
df = df.drop(columns='sports')
```
